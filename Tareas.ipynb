Objetivo: Crear un sistema que:

Encuentre un objeto en una escena (SIFT)
Lo recorte automáticamente (GrabCut)
Sin intervención humana
1. ENTRADA
   ├─ Imagen Template (lo que busco)
   └─ Imagen Escena (donde busco)

2. DETECCIÓN [SIFT]
   ├─ Extraer keypoints de ambas
   ├─ Calcular descriptores
   └─ Matching robusto (Ratio Test)

3. LOCALIZACIÓN [HOMOGRAFÍA]
   ├─ Encontrar transformación geométrica
   ├─ Proyectar esquinas del template
   └─ Calcular Bounding Box

4. REFINAMIENTO [GRABCUT]
   ├─ Usar Bounding Box como ROI
   ├─ Aplicar segmentación precisa
   └─ Generar máscara final

5. SALIDA
   ├─ Objeto extraído sobre fondo negro
   ├─ Objeto sobre fondo blanco (sticker)
   └─ Solo la máscara
